```python
import pdfplumber
from PyPDF2 import PdfReader
import pandas as pd
```

```python
pdf_path = "Lampiran_Data_Bencana_2024.pdf"
```

```python
reader = PdfReader(pdf_path)

all_text = []

for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text()
    all_text.append({
        "page": page_number,
        "text": text
    })

text_df = pd.DataFrame(all_text)
text_df.head()
```

```python
print(text_df.loc[0, "text"][:1500])
```

```python
all_tables = []

with pdfplumber.open(pdf_path) as pdf:
    for page_number, page in enumerate (pdf.pages, start=1):
        tables = page.extract_tables()

        for table_index, table in enumerate(tables, start=1):
            if table and len(table) > 0: 
                df = pd.DataFrame(table) #gapake header kalo iya pake (table[1:], columns=table[0])
                df["page"] = page_number
                df["table_number"] = table_index
                all_tables.append(df)
```

```python
if all_tables:
    tables_df = pd.concat(all_tables, ignore_index=True)
else:
    tables_df = pd.DataFrame()

tables_df.head()
```

```python
tables_df.columns = [str(col).strip() for col in tables_df.columns]
tables_df = tables_df.dropna(how="all")
tables_df = tables_df.reset_index(drop=True)

tables_df.head()
```

```python
page_content = []

with pdfplumber.open(pdf_path) as pdf:
    for page_number, page in enumerate(pdf.pages, start=1):
        text = page.extract_text()
        tables = page.extract_tables()

        page_content.append({
            "page": page_number,
            "text_preview": text[:500] if text else None,
            "num_tables_found": len(tables)
        })

page_summary_df = pd.DataFrame(page_content)
page_summary_df
```

```python
text_df.to_csv("extracted_pdf2024_text.csv", index=False)
tables_df.to_csv("extracted_tables2024_df.csv", index=False)
page_summary_df.to_csv("extracted_page_summary2024.csv", index=False)
```
